# Thumbnail viewer: look at the acquisitions you picked

The **Vendor archive first look** page finds candidate acquisitions at each site and lets you download them as a CSV. This notebook
turns that CSV into a page of thumbnails, so you can see each acquisition over the site before ordering.

**What it does**
1. Reads the CSV from the first-look page. Its name tells the notebook which vendor it is (for example `legion-legion_good_acquisitions.csv`).
2. Looks up each acquisition in the vendor's own catalog, at its site and date, and fetches the vendor's preview image with your
   credentials (`~/code/credentials.ini`).
3. Writes **one private HTML file** with a card per acquisition: the site up close (white outline), Keep and Drop buttons, and a map
   with the whole preview over imagery. A button downloads the acquisitions you kept as a new CSV.

**Keep it private.** Vendor previews need your login and their terms may not allow sharing them, so the page is written to your
ADAPT space, not to the public repo.

| Vendor | Preview | Placement |
|---|---|---|
| Vantor (Legion, WorldView) | `browse` GeoTIFF | exact |
| Planet (SuperDove, Tanager) | `thumbnail` | on the scene's bounding box: close |
| Satellogic | tile `preview`s, pieced together | on each tile's bounding box: close |
| Hydrosat | `THUMBNAIL` | on the scene's bounding box: approximate |
| Kuva Space | none in the catalog | cards show the details only |
